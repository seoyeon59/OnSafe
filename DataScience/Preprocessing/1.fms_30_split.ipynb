```python
# 라이브러리 할당
import os
import cv2
import glob
from tqdm import tqdm
```

```python
# 데이터 가져오기
base_dir = '/content/drive/MyDrive/OnSafe/dataset'    # 기본 디렉토리(폴더) 설정

# 저장할 디렉토리 설정
output_dir = '/content/drive/MyDrive/OnSafe/fms_30_dataset'

if not os.path.exists(output_dir): # 디렉토리 없으면 생성
    os.makedirs(output_dir)
```

```python
# 비디오 분할 함수 정의
def split_video(video_path, output_subdir, frames_per_clip=30):
    video_name = os.path.splitext(os.path.basename(video_path))[0]

    # ✅ 이미 분할된 클립이 있으면 스킵
    first_clip = os.path.join(output_subdir, f"{video_name}_clip_0000.mp4")
    if os.path.exists(first_clip):
        return 0

    cap = cv2.VideoCapture(video_path)

    if not cap.isOpened():
        print(f"{video_name} 파일을 열 수 없습니다.")
        return

    # 영상 정보 읽기
    fps = int(cap.get(cv2.CAP_PROP_FPS))
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    print(f"{video_name} | 총 프레임: {total_frames}, FPS: {fps}, 해상도: {width}x{height}")

    clip_index = 0
    frame_count = 0
    out = None

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        # 30프레임마다 새 클립 생성
        if frame_count % frames_per_clip == 0:
            if out is not None:
                out.release()

            clip_filename = os.path.join(
                output_subdir, f"{video_name}_clip_{clip_index:02d}.mp4"
            )
            out = cv2.VideoWriter(
                clip_filename,
                cv2.VideoWriter_fourcc(*'mp4v'),
                fps,
                (width, height)
            )
            clip_index += 1

        out.write(frame)
        frame_count += 1

    # 마지막 클립 닫기
    if out is not None:
        out.release()
    cap.release()
    print(f"{video_name} 비디오 분할 완료 (총 {clip_index}개 클립 저장)")

    return clip_index
```

```python
# for문을 이용해서 데이터셋 불러오기 및 비디오 분할 함수 적용
labels = ["FALL"]
total_clips = 0

for label in labels:
    input_dir = os.path.join(base_dir, label)
    save_dir = os.path.join(output_dir, label)
    os.makedirs(save_dir, exist_ok=True)

    video_list = glob.glob(os.path.join(input_dir, "*.mp4")) + glob.glob(os.path.join(input_dir, "*.MP4"))
    print(f"\n[{label}] 원본 영상 수: {len(video_list)}")

    for vp in tqdm(video_list, desc=f"Splitting {label}"):
        total_clips += split_video(vp, save_dir, frames_per_clip=30)

print("\n✅ 전체 생성된 클립 수:", total_clips)
print("✅ 저장 위치:", output_dir)
```

```python
# for문을 이용해서 데이터셋 불러오기 및 비디오 분할 함수 적용 - ADL
labels = ["ADL"]
total_clips = 0

for label in labels:
    input_dir = os.path.join(base_dir, label)
    save_dir = os.path.join(output_dir, label)
    os.makedirs(save_dir, exist_ok=True)

    # 하위 폴더까지 모두 검색
    video_list = glob.glob(os.path.join(input_dir, "**", "*.*"), recursive=True)
    # 그 후 폴더가 아닌 파일만 필터링
    video_list = [f for f in video_list if os.path.isfile(f)]
    print(f"\n[{label}] 원본 영상 수: {len(video_list)}")

    for vp in tqdm(video_list, desc=f"Splitting {label}"):
        total_clips += split_video(vp, save_dir, frames_per_clip=30)

print("\n✅ 전체 생성된 클립 수:", total_clips)
print("✅ 저장 위치:", output_dir)
```
